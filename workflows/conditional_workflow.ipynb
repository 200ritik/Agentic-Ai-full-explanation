```python
# conditional workflow
from langgraph.graph import StateGraph, START, END
from typing import TypedDict


# 1. Define state
class TemperatureState(TypedDict):
    temperature: float
    result: str


# 2. Create graph
graph = StateGraph(TemperatureState)


# 3. Node
def check_temperature(state: TemperatureState):
    if state["temperature"] > 30:
        return {"result": "It is hot"}
    else:
        return {"result": "It is normal"}


# 4. Decision function
def decide_temperature(state: TemperatureState):
    if state["temperature"] > 30:
        return "hot"
    else:
        return "normal"


# 5. Add node
graph.add_node("check_temperature", check_temperature)


# 6. START → node
graph.add_edge(START, "check_temperature")


# 7. Conditional edge
graph.add_conditional_edges(
    "check_temperature",
    decide_temperature,
    {
        "hot": END,
        "normal": END
    }
)


# 8. Compile
workflow = graph.compile()


# 9. Run
temperature = float(input("Enter temperature: "))

result = workflow.invoke({
    "temperature": temperature
})

print(result)
```
