```python
from langgraph.graph import START, END, StateGraph
from typing import TypedDict
```

```python
class temp_wth_state(TypedDict):
    celsius_state : float
    fehrenhiet_state : float
    weather_state : str
```

```python
graph = StateGraph(temp_wth_state)
```

```python
def temp_convert(state: temp_wth_state) -> temp_wth_state:
    celsius = state['celsius_state']
    # conversion
    feh = (celsius*9/5) + 32
    state['fehrenhiet_state'] = round(feh, 2)
    
    return state
    
    
    
```

```python
def label_weather(state: temp_wth_state) -> temp_wth_state:
    if (state['fehrenhiet_state'] < 0):
        temp = "extream cold ad even more cold you will die"
        state['weather_state'] = temp
        
    elif(state['fehrenhiet_state']>0 and state['fehrenhiet_state']<50):
        temp = "cold"
        state['weather_state'] = temp
        
    elif(state['fehrenhiet_state']>50 and state['fehrenhiet_state']<70):
        temp = 'mild'
        state['weather_state'] = temp
        
    elif(state['fehrenhiet_state']>70 and state['fehrenhiet_state']<90):
            temp = 'hot'
            state['weather_state'] = temp
            
    elif(state['fehrenhiet_state']>90 and state['fehrenhiet_state']<140):
            temp = 'super hot'
            state['weather_state'] = temp
    
    
    else:
           temp = "vaporizer"
           state['weather_state'] = temp
           
    return state
```

```python
# now the nodes
graph.add_node("temp_convert", temp_convert)
graph.add_node("label_weather", label_weather)
# edges
graph.add_edge(START, "temp_convert")
graph.add_edge("temp_convert", "label_weather")
graph.add_edge("label_weather", END)
```

```python
workflow = graph.compile()


degree =  float(input("enter the degree which you want to convert: "))

# now execution
initial_state= {'celsius_state': degree}
final_state = workflow.invoke(initial_state)

print(final_state)
```
